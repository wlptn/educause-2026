# Lab 01: Text Analysis with LLMs

In this lab you will use an LLM to perform common NLP tasks:
sentiment analysis, summarization, entity extraction, and classification.
No NLP libraries needed -- the model handles everything.

```python
import os
import json
from openai import OpenAI

client = OpenAI(
    base_url=os.environ["MODEL_ENDPOINT"],
    api_key=os.environ["MODEL_API_KEY"],
)
MODEL = os.environ["MODEL_NAME"]
print(f"Model: {MODEL}")
```

## Sentiment Analysis

```python
reviews = [
    "The course material was excellent and the professor explained concepts clearly.",
    "Too much homework and the exams were unfairly difficult.",
    "Decent course overall. Some lectures were engaging, others were dry.",
    "This class changed my perspective on AI. Highly recommend!",
]

for review in reviews:
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": "Classify the sentiment as POSITIVE, NEGATIVE, or NEUTRAL. Reply with only the label."},
            {"role": "user", "content": review},
        ],
        max_tokens=10,
    )
    sentiment = response.choices[0].message.content.strip()
    print(f"{sentiment:10s} | {review[:70]}")
```

## Structured Entity Extraction

Ask the model to extract structured data from unstructured text.

```python
text = """
Dr. Sarah Chen from MIT published a paper on transformer architectures
in Nature Machine Intelligence on March 15, 2026. The research was
funded by a $2.5M grant from the National Science Foundation and
involved collaboration with researchers at Stanford and Oxford.
"""

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "Extract entities from the text. Return valid JSON with keys: people, organizations, dates, monetary_values, topics."},
        {"role": "user", "content": text},
    ],
)

print(response.choices[0].message.content)
```

## Summarization at Different Levels

```python
article = """
Artificial intelligence is transforming higher education in several ways.
Universities are adopting AI-powered tutoring systems that provide
personalized feedback to students. Professors are using AI to automate
grading of assignments, freeing up time for research and mentorship.
Administrative departments leverage AI for enrollment prediction and
resource allocation. However, there are concerns about academic integrity,
as students may use AI to complete assignments. Institutions are developing
new policies to address these challenges while embracing the benefits.
The key is finding a balance between leveraging AI as a learning tool
and maintaining rigorous academic standards.
"""

audiences = [
    "a university president (1 sentence, focus on strategy)",
    "a first-year student (2 sentences, simple language)",
    "a faculty senate (3 bullet points, focus on policy implications)",
]

for audience in audiences:
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": f"Summarize the following text for {audience}."},
            {"role": "user", "content": article},
        ],
    )
    print(f"\nFor {audience.split('(')[0].strip()}:")
    print(f"  {response.choices[0].message.content.strip()}")
    print("-" * 60)
```
