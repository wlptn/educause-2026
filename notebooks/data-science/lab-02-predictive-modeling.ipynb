```python
import os
import pandas as pd
import numpy as np
from openai import OpenAI

client = OpenAI(
    base_url=os.environ["MODEL_ENDPOINT"],
    api_key=os.environ["MODEL_API_KEY"],
)
MODEL = os.environ["MODEL_NAME"]

np.random.seed(42)
n = 300
df = pd.DataFrame({
    "sq_feet": np.random.normal(1500, 400, n).clip(500, 4000).astype(int),
    "bedrooms": np.random.choice([1, 2, 3, 4, 5], n, p=[0.1, 0.25, 0.35, 0.2, 0.1]),
    "age_years": np.random.exponential(15, n).clip(0, 80).astype(int),
    "has_garage": np.random.choice([0, 1], n, p=[0.3, 0.7]),
})
df["price"] = (
    50000 + 120 * df["sq_feet"] + 15000 * df["bedrooms"] -
    800 * df["age_years"] + 25000 * df["has_garage"] +
    np.random.normal(0, 20000, n)
).clip(50000).astype(int)

print(f"Dataset: {len(df)} homes")
df.head()
```

## Ask the AI for modeling advice

```python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a data science instructor. Give practical, actionable advice."},
        {"role": "user", "content": f"""I have a housing dataset with columns: {list(df.columns)}.
I want to predict 'price'. Here are the first few rows:

{df.head().to_string()}

What modeling approach would you recommend for a beginner? Walk me through it step by step.
Include the Python code I should run."""}
    ],
)

print(response.choices[0].message.content)
```
