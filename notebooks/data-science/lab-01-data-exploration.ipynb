```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from openai import OpenAI

client = OpenAI(
    base_url=os.environ["MODEL_ENDPOINT"],
    api_key=os.environ["MODEL_API_KEY"],
)
MODEL = os.environ["MODEL_NAME"]

print(f"Model: {MODEL}")
print("Ready to explore data with AI assistance.")
```

## Generate a sample dataset

```python
np.random.seed(42)
n = 200
df = pd.DataFrame({
    "student_id": range(1, n + 1),
    "study_hours": np.random.normal(15, 5, n).clip(0, 40),
    "sleep_hours": np.random.normal(7, 1.5, n).clip(3, 12),
    "attendance_pct": np.random.normal(80, 15, n).clip(0, 100),
})
df["exam_score"] = (
    0.3 * df["study_hours"] + 0.2 * df["sleep_hours"] +
    0.4 * df["attendance_pct"] / 100 * 100 +
    np.random.normal(0, 8, n)
).clip(0, 100).round(1)

df.head(10)
```

## Ask the AI to analyze the data

Provide the model with a summary of the dataset and ask for analysis guidance.

```python
summary = df.describe().to_string()
correlations = df.corr(numeric_only=True).to_string()

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a data science teaching assistant. Be concise and practical."},
        {"role": "user", "content": f"""Here is a dataset of student performance.

Summary statistics:\n{summary}

Correlations:\n{correlations}

What are the 3 most interesting findings? Suggest a visualization for each."""}
    ],
)

print(response.choices[0].message.content)
```
