# Lab 01: Your First LLM API Call

In this lab you will make your first call to a large language model using
the OpenAI-compatible API. The endpoint and API key are already configured
in your environment -- no setup required.

```python
# One-time setup: install the Python packages this lab uses
%pip install -q openai
```

```python
import os
from openai import OpenAI

client = OpenAI(
    base_url=os.environ["MODEL_ENDPOINT"],
    api_key=os.environ["MODEL_API_KEY"],
)

response = client.chat.completions.create(
    model=os.environ["MODEL_NAME"],
    messages=[
        {"role": "user", "content": "Explain what a large language model is in two sentences."}
    ],
)

print(response.choices[0].message.content)
```

## How it works

- `MODEL_ENDPOINT` points to a **Models-as-a-Service gateway** that routes
  requests to the right backend (could be on-cluster, Azure, AWS, etc.)
- `MODEL_API_KEY` authenticates your request
- `MODEL_NAME` selects which model to use

Your IT admin provisioned all of this through **Ansible Automation Platform**
when they created your course environment.

## Exercise 2: Adjust temperature

Temperature controls randomness. Try values between 0.0 (deterministic) and
1.0 (creative). Run the cell multiple times at each setting and observe how
the output changes.

```python
for temp in [0.0, 0.5, 1.0]:
    response = client.chat.completions.create(
        model=os.environ["MODEL_NAME"],
        temperature=temp,
        messages=[
            {"role": "user", "content": "Write a one-sentence summary of machine learning."}
        ],
    )
    print(f"temp={temp}: {response.choices[0].message.content}\n")
```

## Exercise 3: System messages

A `system` message gives the model a persona or instructions that shape
all subsequent responses.

```python
response = client.chat.completions.create(
    model=os.environ["MODEL_NAME"],
    messages=[
        {"role": "system", "content": "You are a helpful teaching assistant for an introductory AI course. Keep answers concise and beginner-friendly."},
        {"role": "user", "content": "What is the difference between supervised and unsupervised learning?"}
    ],
)

print(response.choices[0].message.content)
```
